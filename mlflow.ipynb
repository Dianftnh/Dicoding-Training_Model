```python
import os
import mlflow
import shutil # Import shutil for rmtree

os.environ["MLFLOW_ALLOW_FILE_STORE"] = "true"

TRACKING_URI = os.path.join(os.getcwd(), "mlruns")
trash_path = os.path.join(TRACKING_URI, ".trash")

# Create the tracking directory if it does not exist
os.makedirs(TRACKING_URI, exist_ok=True)

# Remove .trash if it exists and then recreate it, to ensure it's a clean directory
if os.path.exists(trash_path):
    if os.path.isdir(trash_path):
        shutil.rmtree(trash_path)
    else:
        os.remove(trash_path)

os.makedirs(trash_path, exist_ok=True)

mlflow.set_tracking_uri(f"file:{TRACKING_URI}")
mlflow.set_experiment("imdb-review-tuning")
```
